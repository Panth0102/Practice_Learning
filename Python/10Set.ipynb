**Add- Remove - Update** <br>
Set only has unique values, also it is always sorted <br>
Add: Adds a new value
Remove: Removes the exact element from the set
Update: If string passed then individual characters are saved, else if iterable are passed than values would be insereted

```python
set1 = {"Apple", "Banana", "Chikoo"}

set1.add("DragonFruit")

set1.remove("Apple")

set1.update("Eberry")

set1.update(["Franberry", "Guava", "Hemon"])

set1.update(["Apple"])
set1
```

**Basic Set Functionalities**

```python
data1 = {"ram","seeta","lakshman","kush","luv","krishna"}
data2 = {"ram","arjunn","bhim","sahdeve","krishna","draupadi"}

# print(data1)
# print(data2)

#x  = data1.union(data2)
x = data1 | data2
print(x)

#x = data1.intersection(data2)
x = data1 & data2
print(x)

#x = data1.difference(data2)
x = data2 -data1
print(x)

x = data1.symmetric_difference(data2)
print(x)

y = data1.issuperset(data2)
print(y)

y = data2.issubset(data1)
print(y)
```

```python
data1.intersection_update(data2)
print(data1)
```
