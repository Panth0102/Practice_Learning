**For Loop**

```python
for i in range(50, 30, -2):
  print(i)

cities = []
n = int(input("How many cities do you want to insert?\n"))
for i in range(n):
  city = input(f"city-{i+1}: ")
  cities.append(city)
print(cities)
```

**While**

```python
temp = []
while True:
  a = input("Enter a new member of the list (or enter 'x' to exit): ")
  if a == "x":
    break
  temp.append(a)
temp
```
