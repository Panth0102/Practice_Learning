**Basic Class**

```python
class Car:
    def __init__(self, model, speed, price):
        self.model = model
        self.speed = speed
        self.price = price

    def carDetails(self):
        print("Car is " + self.model + " with top speed of " + str(self.speed) + " km/hr at price of " + str(self.price))

    def carDetailsInFString(self):
        print(f"Car is {self.model} with top speed of {self.speed} km/hr at price of {self.price}")
```

**Basic Object to use method with constructor**

```python
car1 = Car("BMW M4", 270, 15000000)
car1.carDetails()
car1.carDetailsInFString()
```

**Inheritance**

**Simple Inheritance**

```python
class Engine(Car):
    def __init__(self, model, speed, price, engine, petrolCapacity):
        super().__init__(model, speed, price)
        self.engine = engine
        self.petrolCapacity = petrolCapacity

    def detailedVechile(self):
        self.carDetails()
        print(f"The car has a {self.engine} engine with {self.petrolCapacity} litres of tank")
```

```python
car2 = Engine("BMW M5", 300, 17500000, "V12", 200)
car2.carDetailsInFString()
car2.detailedVechile()
```

**Hierarchical Inheritance**

```python
class MusicSystem(Car):
    def __init__(self, model, speed, price, system, volume):
        super().__init__(model, speed, price)
        self.system = system
        self.volume = volume

    def musicVechile(self):
        self.carDetailsInFString()
        print(f"The car has a {self.system} audio system and max volume is {self.volume}")
```

```python
car3 = MusicSystem("BMW M8", 350, 20000000, "Dolby", 42)
car3.musicVechile()
```

**Multiple and Multilevel**
Must use **kwargs method to use Multiple and Multi Level inheritance

```python
class Car:
    def __init__(self, model, speed, price, **kwargs):
        super().__init__(**kwargs)
        self.model = model
        self.speed = speed
        self.price = price

    def carDetails(self):
        print(f"Car is {self.model} with top speed of {self.speed} km/hr at price of {self.price}")


class Engine(Car):
    def __init__(self, engine, petrolCapacity, **kwargs):
        super().__init__(**kwargs)
        self.engine = engine
        self.petrolCapacity = petrolCapacity

    def engineDetails(self):
        print(f"Engine : {self.engine}")
        print(f"Petrol Capacity : {self.petrolCapacity} litres")


class MusicSystem(Car):
    def __init__(self, system, volume, **kwargs):
        super().__init__(**kwargs)
        self.system = system
        self.volume = volume

    def musicDetails(self):
        print(f"Music System : {self.system}")
        print(f"Maximum Volume : {self.volume}")


class LuxuryCar(MusicSystem, Engine):
    def __init__(self, model, speed, price,engine, petrolCapacity,system, volume,carClass):

        super().__init__(model=model,speed=speed,price=price,engine=engine,petrolCapacity=petrolCapacity,system=system,volume=volume)

        self.carClass = carClass

    def luxuryDetails(self):
        self.carDetails()
        self.engineDetails()
        self.musicDetails()
        print(f"Class : {self.carClass}")


car = LuxuryCar(model="BMW M5",speed=305,price=15000000,engine="V8 Twin Turbo",petrolCapacity=68,system="Bose",volume=100,carClass="Luxury Sedan")

car.luxuryDetails()
```
