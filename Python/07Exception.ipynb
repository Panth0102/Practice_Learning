```python
try:
    no1 = int(input("enter no 1:"))
    no2 = int(input("enter no 2:"))
    ans = no1 / no2 #exception ... except block
    print("ans = ",ans)
except ZeroDivisionError:
    print("can not divide by zero") 
except ValueError as e:
    print("value error..",e)    
```

```python
name = input("Enter a name: ")

try:
    if name != name[::-1]:
        raise StringError("name is not plaindrome...")
    else:
        print("It is palindrome")
except StringError as e:
    print(e)    
except ValueError:
    print("value error...")    
```

```python
email = "ram_gmail.com"

try:
    if "@" not in email:
        raise StringError("email is in valid..")
    print("email ",email)
except StringError as e:
    print(e)    
```
