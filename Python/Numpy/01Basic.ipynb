```python
import time
size = 100000

start = time.time()
basic_list = list(range(size))
res = [x+1 for x in basic_list]
time_python = time.time() - start
print(f"Basic List time is {time_python:.6f}")


start = time.time()
numpy_list = np.arange(size)
res = numpy_list + 1
time_python = time.time() - start
print(f"Numpy List time is {time_python:.6f}")
```

**2.Creating numphy Array**


```python
# List method
np_arr1 = np.array([1,2,3,4,5])

# Arange        start,stop,steps
np_arr2 = np.arange(0,10,2)

#linspace       start, stop, parts
np_arr3 = np.linspace(0,1,10)

# Aray of Zero
np_arr4 = np.zeros(5)

#Array of ones
np_arr5 = np.ones(5)

# Full array   size, fill num
np_arr6 = np.full(5,10)

# Identity matrix
np_arr7 = np.eye(3)

# Randon
np_arr8 = np.random.rand(10)

# Random integer        start, stop, size
np_arr9 = np.random.randint(0,10,size=5)



print(np_arr1)
print(np_arr2)
print(np_arr3)
print(np_arr4)
print(np_arr5)
print(np_arr6)
print(np_arr7)
print(np_arr8)
print(np_arr9)
```

**3.Datatypes or dtypes**

```python
arr_int = np.array([1,2,3])
arr_flo = np.array([1.,2.,3.])
arr_mix = np.array([1.0,2,3])

print(arr_int.dtype)
print(arr_flo.dtype)
print(arr_mix.dtype)

arr_defined = np.array([1,2,3], dtype='float32')
print(arr_defined.dtype)

arr_new = arr_int.astype('float32')
print(arr_new.dtype)
```

**5. Indexing & Slicing**

```python
np_arr = np.random.randint(0,100, size= 10)
```

```python
print(np_arr)
print(np_arr[0])
print(np_arr[-1])
print(np_arr[::-1])
```

```python
arr = np.random.randint(0,100, size= (4,4))

arr
```

```python
print(arr[0,0])             #Prints a specific element
print(arr[0, 0:])           #Prints only column
print(arr[0: , 0])          #Prints only row
print(arr[0: , 0:])         #Prints all elements
print(arr[1:3, 1:3])       #Prints a specific part of the array
```

```python
print(np_arr[np_arr > 50])   #Prints all elements greater than 50
print(np_arr[np_arr % 2 == 0])   #Prints all elements divisble by 2
print(arr[0: , 0:][arr[0: , 0:] > 50])   #Prints all elements greater than 50
```

**9. Linear Algebra**

```python
a = np.random.randint(0,10,size=(2,2))
b = np.random.randint(0,10,size=(2,2))

print(a)
print(b)
print(f'Linear Algebra {a*b}')
print(f'Dot multiplication {np.dot(a,b)}')      #Matrix multiplication

print(f'Determinant : {np.linalg.det(a)}')
print(f'Inverse: {np.linalg.inv(a)}')
print(f'Transpose: {a.T}')

eigenvals, eigenvec = np.linalg.eigh(a)
print(f'Eigen Values : {eigenvals}')
print(f'Eigen Vectors : {eigenvec}')
```

**10.Useful Array Methods**

```python
a = np.random.randint(0,10,size=15)
b = np.random.randint(0,10,size=15)

#Sorting

print(f'Original Value : {a}')
print(f'Sorts and produces a copy : {np.sort(a)}')
print(f'Does not change the original array : {a}')

a.sort()
print(f'Sorts the original array : {a}')

#Unique

print(f'Unique values : {np.unique(a)}')


a = np.random.randint(0,10,size=15)
```

**11. How can numphy be used for image**

```python
img = np.random.randint(0,256,size=(10,10))
print(img)
```

```python
print('Image statsics')
print(f'Brightness: {np.mean(img)}')
print(f'Contrast: {np.std(img)}')
print(f'Min pixel: {np.min(img)}')
print(f'Max pixel: {np.max(img)}')

```

```python
brighter = np.clip(img + 50,0,255)
print(f'Old Brightness : {np.mean(img)}')
print(f'New Brightness : {np.mean(brighter)}')
```

```python
contrast = np.clip(img*1.5, 0,255)
print(f'Old Contrast : {np.std(img)}')
print(f'New Contrast : {np.std(brighter)}')
```
