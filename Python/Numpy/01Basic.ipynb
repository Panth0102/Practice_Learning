**1.Why Use Numphy**

```python
np_arr = np.array([1,2,3,4,5])

np_arr
np_arr+10       #array([11, 12, 13, 14, 15])
np_arr          #array([1, 2, 3, 4, 5])
type(np_arr)    #np_arr = np.array([1,2,3,4,5])
```

```python
import time
size = 100000

start = time.time()
basic_list = list(range(size))
res = [x+1 for x in basic_list]
time_python = time.time() - start
print(f"Basic List time is {time_python:.6f}")


start = time.time()
numpy_list = np.arange(size)
res = numpy_list + 1
time_python = time.time() - start
print(f"Numpy List time is {time_python:.6f}")
```

**2.Creating numphy Array**


```python
# List method
np_arr1 = np.array([1,2,3,4,5])

# Arange        start,stop,steps
np_arr2 = np.arange(0,10,2)

#linspace       start, stop, parts
np_arr3 = np.linspace(0,1,10)

# Aray of Zero
np_arr4 = np.zeros(5)

#Array of ones
np_arr5 = np.ones(5)

# Full array   size, fill num
np_arr6 = np.full(5,10)

# Identity matrix
np_arr7 = np.eye(3)

# Randon
np_arr8 = np.random.rand(10)

# Random integer        start, stop, size
np_arr9 = np.random.randint(0,10,size=5)



print(np_arr1)
print(np_arr2)
print(np_arr3)
print(np_arr4)
print(np_arr5)
print(np_arr6)
print(np_arr7)
print(np_arr8)
print(np_arr9)
```
