**Compile Time - Static Polymorphism**

Compile Time / static polymorphism refers as method overloading, meaning more than one function with same name but different parameters in a class, but in Python it is not supported.

```python
class Bank:
    
    def __init__(self):
        print("bank class const called...")
    
    def deposit(self,a,b):
        print("deposit with double argument called..",a,b)        
        
    def deposit(self,a):
        print("deposit with single argument called..",a)    


b = Bank()
# b.deposit(10,20)   Genertes Error
b.deposit(10)   # The first method is overwritten by the second
```

Third-party dispatch library

```python

from multipledispatch import dispatch

class Bank:
    
    def __init__(self):
        print("bank class const called...")
    
    @dispatch(int,int)
    def deposit(self,a,b):
        print("deposit with double argument called..",a,b)        
        
    @dispatch(int)    
    def deposit(self,a):
        print("deposit with single argument called..",a)    


b = Bank()
b.deposit(10,20)        
```

**Run Time - Dynamic Polymorphism**

Run Time/ Dynamic Polymorphism refers as method overriding, meaning a derived calss and base class has same method, and only one method can be used

```python
class RBI:
    
    def __init__(self):
        pass
    
    
    def withdraw(self):
        print("withdraw from rbi")


class SBI(RBI):
    
    def __init__(self):
        pass
    
    
    def withdraw(self):
        print("withdraw from SBI")

s = SBI()
s.withdraw()        
```

For a abstract method, it is compulsory to define it within the base class

```python
from abc import ABC,abstractmethod
#abc -->pakcage
#ABC class
#abstract base class

#abstract class
class RBI(ABC):
    
    def __init__(self):
        pass
    
    
    @abstractmethod
    def withdraw(self):
        pass
    
    def loan(self):
        pass


class SBI(RBI):
    
    def __init__(self):
        pass
    
    
    def withdraw(self):
        print("withdraw from SBI")

s = SBI()
s.withdraw()        
```

**Operator Overloading**

```python
class Cart:
    
    def __init__(self,items):
        self.items = items
    
    # + op overload
    def __add__(self, other):
        #self -->left side op --> other -->right side op
        print("self",self.items)
        print("other",other.items)
        return Cart(self.items + other.items)


cart1 =Cart([{"name":"iphone","price":1200}])        
cart2 =Cart([{"name":"ipad","price":1400}])        


# + -> __add__()
final = cart1 + cart2
print(final.items)
```
