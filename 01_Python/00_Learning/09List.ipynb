**Append - Insert - Extend** <br>
Append: Adds at the last of the list <br>
Insert: Adds at a given index <br>
Extend: Adds a whole new list to an exisitng list, if string passed, divides it into smaller portion

```python
users = ["Ram", "Shyam", "Balram"]


users.append("Abhishek")

users.insert(1,"Balram")

users.extend("Param")

users
```

**Pop - Remove** <br>
Pop: Removes the last value added or at given index <br>
Remove: Removes a specific value

```python
users.pop()

users.pop(4)

users.remove("Balram")

users
```

**Unique**

```python
colors = ["red","green","yellow","yellow","yellow","pink","white","white","black"]
#unique list

uniqueList = []
duplicate=[]

for i in colors:
    if i not in uniqueList:
        uniqueList.append(i) #red
    else:
        duplicate.append(i)    
print(uniqueList)        
print(duplicate)
```

**Index - Count - Reverse - Sort** <br>
Index: Gives first occurence of a value <br>
Count: Gives total count of a value <br>
Marks: Reverses a list <br>
Sort: Sorts the list


```python
marks = [10,20,23,12,14,12]
print(marks)

ind = marks.index(12)
print("index ",ind)

cnt =marks.count(12)
print("cnt",cnt)

marks.reverse()
print(marks)

marks.sort()
marks.sort(reverse=True)
marks
```
