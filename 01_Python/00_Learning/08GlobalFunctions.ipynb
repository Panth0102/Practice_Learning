**Sorted**

```python
nums = [5,2,1,8]
#sortlist =sorted(nums)
sortlist =sorted(nums,reverse=True)
print(sortlist)


users = ["amit","sumit","raj","parth"]
#sortedUsers = sorted(users)
#sortedUsers = sorted(users,reverse=True)
sortedUsers = sorted(users,key=len)

print(sortedUsers)

nums = [5,2,-1,8,-20]
sortlist =sorted(nums)
print(sortlist)
```
