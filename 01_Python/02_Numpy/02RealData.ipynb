```python
import numpy as np
```

```python
data = np.genfromtxt(
    '../Database/employees.csv',
    delimiter=',',
    names=True,
    dtype=[
        ('EMPLOYEE_ID', 'i4'),
        ('FIRST_NAME', 'U20'),
        ('LAST_NAME', 'U20'),
        ('EMAIL', 'U30'),
        ('PHONE_NUMBER', 'U20'),
        ('HIRE_DATE', 'U20'),
        ('JOB_ID', 'U20'),
        ('SALARY', 'f8'),
        ('COMMISSION_PCT', 'i4'),
        ('MANAGER_ID', 'i4'),
        ('DEPARTMENT_ID', 'i4')
    ],
    missing_values=['-', ' - '],
    filling_values=0,
    encoding='utf-8'
)
```

```python
print(f'Shapes : {data.shape}')
print(f'Columns : {data.dtype.names}')
print(f'First 5 rows : {data[:5]}')
print(f'Total rows : {data.shape[0]}')
print(f'Data Dtype : {data.dtype}')
print(f'Data at 0 : {data[0]}')
print(f'Data at 0,0 : {data[0][0]}')
```

```python
print(f'Average Salary : {np.mean(data["SALARY"])}')
print(f'Salaray with department_id 10 is : {np.mean(data["SALARY"][data["DEPARTMENT_ID"] == 10])}')
```

```python
data["SALARY"] = np.where(data["DEPARTMENT_ID"] == 10, data["SALARY"]*1.1, data["SALARY"])
print(f'Average Salary : {np.mean(data["SALARY"])}')
```

```python
print(f'Number of employees in the company: {np.unique(data["EMPLOYEE_ID"]).size}')
print(f'Number of employees in each department: {np.unique(data["DEPARTMENT_ID"], return_counts=True)}')
print(f'Maximum employees are in department: {np.unique(data["DEPARTMENT_ID"], return_counts=True)[0][np.argmax(np.unique(data["DEPARTMENT_ID"], return_counts=True)[1])]}')
print(f'Minimum employees are in department: {np.unique(data["DEPARTMENT_ID"], return_counts=True)[0][np.argmin(np.unique(data["DEPARTMENT_ID"], return_counts=True)[1])]}')
print(f'Unique job ids are: {np.unique(data["JOB_ID"])}')
print(f'Number of job titles is: {np.unique(data["JOB_ID"]).size}')
print(f'Minimum employees under a managar is: {np.unique(data["MANAGER_ID"], return_counts=True)[0][np.argmin(np.unique(data["MANAGER_ID"], return_counts=True)[1])]}')
print(f'Maximum employees under a managar is: {np.unique(data["MANAGER_ID"], return_counts=True)[0][np.argmax(np.unique(data["MANAGER_ID"], return_counts=True)[1])]}')

years = np.array([date[-2:] for date in data["HIRE_DATE"]])
unique_years, counts = np.unique(years, return_counts=True)

print(unique_years)
print(counts)
```

**Must do questions**

Level 1 — Basic

1. Which department has the most employees?
2. Which department has the highest average salary?
3. Who is the highest-paid employee?
4. What is the average company salary?
5. Which job role has the most employees?



```python
print(f'Maximum employees are in department: {np.unique(data["DEPARTMENT_ID"], return_counts=True)[0][np.argmax(np.unique(data["DEPARTMENT_ID"], return_counts=True)[1])]}')
print()

dept_ids = np.unique(data["DEPARTMENT_ID"])
total_salary = np.bincount(
    data["DEPARTMENT_ID"],
    weights=data["SALARY"]
)
employee_count = np.bincount(data["DEPARTMENT_ID"])
average_salary = np.divide(total_salary, employee_count, out=np.zeros_like(total_salary), where=employee_count != 0)

for dept in dept_ids:
    print(f"Department {dept}: {average_salary[dept]:.2f}")
print()

print(f'Maximum salary is : {np.max(data["SALARY"])} and paid to employee id : {data["EMPLOYEE_ID"][np.argmax(data["SALARY"])]}')
print()

print(f'Average company salary is : {np.mean(data["SALARY"])}')
print()

print(f'Job with highest employees is : {np.unique(data["JOB_ID"], return_counts=True)[0][np.argmax(np.unique(data["JOB_ID"], return_counts=True)[1])]}')
```
